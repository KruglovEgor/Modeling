<h2 style="text-align: center;"><b>Введение</b></h2>

Любой алгоритм машинного обучения с учителем, решающий задачу *классификации* или *регрессии*, включает три стадии:

1. *Стадия инициализации*: задаются **гиперпараметры**, то есть те величины, которые не "выучиваются" алгоритмом в процессе обучения самостоятельно, но влияют на качество его работы.
2. *Стадия обучения*: алгоритм запускается на данных, **обучаясь** на них и меняя свои **параметры** (не путать с *гипер*параметрами) так, чтобы минимизировать функцию потерь (*loss function*), например, с помощью *метода градиентного спуска* или *метода коррекции ошибки*. Функция потерь, по сути, показывает, где и как ошибается модель.
3.  *Стадия предсказания*: модель обучена, и теперь с помощью неё можно делать **предсказания** на новых объектах.

<h2 style="text-align: center;"><b>Класс Perceptron</b></h2>

В даном разделе с помощью перцептрона решается задача **бинарной классификации**:  
- *Входные данные*: матрица $X$ размера $(n, m)$ и вектор (столбец) $y$ из нулей и единиц размера $(n, 1)$. Строкам матрицы $X$ соответствуют объекты, столбцам - признаки (то есть строка $i$ – это набор признаков (*признаковое описание*) объекта $X_i$).
- *Выходные данные*: вектор $\hat{y}$ из нулей и единиц размера $(n, 1)$ – предсказания алгоритма.

Схема искусственного нейрона:  

<img src="https://neerc.ifmo.ru/wiki/images/a/a5/%D0%98%D1%81%D0%BA%D1%83%D1%81%D1%81%D1%82%D0%B2%D0%B5%D0%BD%D0%BD%D1%8B%D0%B9_%D0%BD%D0%B5%D0%B9%D1%80%D0%BE%D0%BD_%D1%81%D1%85%D0%B5%D0%BC%D0%B0.png"/>

Чтобы понять, как будут обновляться параметры (веса) модели, нужно знать, какая функция потерь оптимизируеся (находится ее минимум). В данном случае решается задача бинарной классификации (2 класса: 1 или 0), возьмём в качестве функции потерь среднеквадратичную ошибку:  

$$Loss(\hat{y}, y) = \frac{1}{2n}\sum_{i=1}^{n} (\hat{y_i} - y_i)^2 = \frac{1}{2n}\sum_{i=1}^{n} (f(w \cdot X_i + b) - y_i)^2$$  

Здесь $w \cdot X_i$ - скалярное произведение, а $f()$ - пороговая функция активации:  

$$
f(w \cdot X_i + b) =
\begin{cases}
1, &\text{если } w \cdot X_i + b > 0 \\
0, &\text{если } w \cdot X_i + b \le 0
\end{cases}
$$  

**Примечание:** На самом деле можно считать, что $b$ - свободный член - является частью вектора весов: $w_0$, приписав к $X$ слева единичный столбец, тогда в скалярном произведении с каждым объектом $b$ будет именно как свободный член. При реализации класса `Perceptron()` будем обновлять $b$ отдельно от $w$.

Реализуем функцию потерь:

```python
def Loss(y_pred, y):
    '''
    Считаем среднеквадратичную ошибку
    '''
    y_pred = y_pred.reshape(-1,1)
    y = np.array(y).reshape(-1,1)
    return 0.5 * np.mean((y_pred - y) ** 2)
```

Поскольку у *пороговой функции* не существует производной в нуле, использовать градиентный спуск нельзя, ведь:  

$$ \frac{\partial Loss}{\partial w} = \frac{1}{n} X^T\left(f(w \cdot X) - y\right)f'(w \cdot X),$$  

где $f^{'}(w \cdot X)$ – в точке 0 посчитать не получится.

Но для обучения модели веса обновлять надо.

Предлагается для обновления весов использовать следующее правило:   

$$w^{j+1} = w^{j} - \alpha\Delta{w^{j}},$$
$$b^{j+1} = b^{j} - \alpha\Delta{b^{j}},$$

где

$$\Delta{w} = \frac{1}{n}X^T(\hat{y} - y) = \frac{1}{n}X^T(f(X \cdot w^j + b^j) - y),$$  
$$\Delta{b} = \frac{1}{n}1^T(\hat{y} - y) = \frac{1}{n}1^T(f(X \cdot w^j + b^j) - y)$$  

Здесь $w \cdot X$ - матричное произведение столбца весов $w$ на матрицу объектов-признаков $X$, $1^T$ – вектор-строка из единиц, индекс $j$ – номер итерации градиентного спуска.

Это правило является частным случаем градиентного спуска для данного случая (см. *[правило Хебба](https://ru.wikipedia.org/wiki/%D0%94%D0%B5%D0%BB%D1%8C%D1%82%D0%B0-%D0%BF%D1%80%D0%B0%D0%B2%D0%B8%D0%BB%D0%BE)*, *[метод коррекции ошибки](https://ru.wikipedia.org/wiki/%D0%9C%D0%B5%D1%82%D0%BE%D0%B4_%D0%BA%D0%BE%D1%80%D1%80%D0%B5%D0%BA%D1%86%D0%B8%D0%B8_%D0%BE%D1%88%D0%B8%D0%B1%D0%BA%D0%B8)*).

Вооружившись всеми формулами, напишем свой класс **`Perceptron()`**.

*Примечание*: В коде ниже `y_pred` – это $\hat{y}$ из формул выше

```python
class Perceptron:
    def __init__(self, w=None, b=0):
        """
        :param: w -- вектор весов
        :param: b -- смещение
        """
        # Пока мы не знаем размер матрицы X, а значит не знаем, сколько будет весов
        self.w = w
        self.b = b

    def activate(self, x):
        return np.array(x > 0, dtype=np.int64) # пороговая функция активации

    def forward_pass(self, X):
        """
        Эта функция рассчитывает ответ перцептрона при предъявлении набора объектов
        :param: X – матрица объектов размера (n, m), каждая строка - отдельный объект
        :return: вектор размера (n, 1) из нулей и единиц с ответами перцептрона
        """
        n = X.shape[0]
        y_pred = np.zeros((n, 1))  # y_pred == y_predicted - предсказанные классы
        y_pred = self.activate(X @ self.w.reshape(X.shape[1], 1) + self.b)
        return y_pred.reshape(-1, 1)

    def backward_pass(self, X, y, y_pred, learning_rate=0.005):
        """
        Функция обновляет значения весов перцептрона в соответствии с объектом
        :param: X -- матрица объектов размера (n, m)
                y -- вектор правильных ответов размера (n, 1)
                learning_rate - "скорость обучения" (символ alpha в формулах выше)
        В этом методе ничего возвращать не нужно, только правильно поменять веса
        с помощью градиентного спуска.
        """
        n = len(y)
        y = np.array(y).reshape(-1, 1)
        self.w = self.w - learning_rate * (X.T @ (y_pred - y) / n)
        self.b = self.b - learning_rate * np.mean(y_pred - y)

    def fit(self, X, y, num_epochs=300):
        """
        Спускаемся в минимум
        :param: X -- матрица объектов размера (n, m)
                y -- вектор правильных ответов размера (n, 1)
                num_epochs -- количество итераций обучения
        :return: losses -- вектор значений функции потерь
        """
        self.w = np.zeros((X.shape[1], 1))  # столбец (m, 1)
        self.b = 0  # смещение
        losses = []  # значения функции потерь на различных итерациях обновления весов

        for i in range(num_epochs):
            # предсказания с текущими весами
            y_pred = self.forward_pass(X)
            # считаем функцию потерь с текущими весами
            losses.append(Loss(y_pred, y))
            # обновляем веса в соответсвие с тем, где ошиблись раньше
            self.backward_pass(X, y, y_pred)

        return losses
```

Класс готов. Посмотрим, правильно ли ведёт себя перцептрон. Для этого запустите ячейки с тестовым кодом и проверьте, чтобы результаты запуска совпадали с соответствующими числами из таблиц:

```python
w = np.array([1., 2.]).reshape(2, 1)
b = 2.
X = np.array([[1., 2., -1.], [3., 4., -3.2]])

perceptron = Perceptron(w, b)
y_pred = perceptron.forward_pass(X.T)
print ("y_pred = " + str(y_pred))
```

Обозначим, что здесь признаки, а что - классы:

```python
X = data.iloc[:,:2].values  # матрица объекты-признаки
y = data['target'].values.reshape((-1, 1))  # классы (столбец из нулей и единиц)
```

<h3 style="text-align: center;"><b>Предсказание пола по голосу</b></h3>

Сравним качество работы нашего перцептрона и алгоритма из библиотеки `sklearn` на датасете с сайта [Kaggle](https://www.kaggle.com) - [Gender Recognition by Voice](https://www.kaggle.com/primaryobjects/voicegender). В данном датасете в качестве признаков выступают различные звуковые характеристики голоса, а в качестве классов - пол (мужчина/женщина). Подробнее о самих признаках можно почитать [на странице датасета](https://www.kaggle.com/primaryobjects/voicegender) (на английском). Нашей целью является просто протестировать на этих данных два алгоритма.

```python
import pandas as pd
from sklearn.linear_model import Perceptron as skPerceptron
from sklearn.metrics import accuracy_score
```

```python
data_path = '/content/voice.csv'
data = pd.read_csv(data_path)
data['label'] = data['label'].apply(lambda x: 1 if x == 'male' else 0)
```

```python
data.head()
```

```python
# Перемешиваем данные. Изначально сначала идут мужчины, потом женщины
data = data.sample(frac=1)
```

```python
X_train = data.iloc[:int(len(data)*0.7), :-1]  # матрица объекты-признаки
y_train = data.iloc[:int(len(data)*0.7), -1]  # истинные значения пола (мужчина/женщина)

X_test = data.iloc[int(len(data)*0.7):, :-1]  # матрица объекты-признаки
y_test = data.iloc[int(len(data)*0.7):, -1]  # истинные значения пола (мужчина/женщина)
```

Натренируем наш перцептрон и перцептрон из `sklearn` на этих данных:

```python
RANDOM_SEED = 42

perceptron = Perceptron()
perceptron.fit(X_train.values, y_train.values)

sk_perceptron = skPerceptron(random_state=RANDOM_SEED)
sk_perceptron.fit(X_train.values, y_train.values)
```

Сравним доли правильных ответов (на тестовых данных):

```python
print('Точность (доля правильных ответов, из 100%) собственного перцептрона: {:.3f} %'.format(
    accuracy_score(y_test.values, perceptron.forward_pass(X_test)) * 100))
print('Точность (доля правильных ответов) перцептрона из sklearn: {:.3f} %'.format(
    accuracy_score(y_test.values, sk_perceptron.predict(X_test)) * 100))
```

Попробуем поставить число итераций побольше:

```python
RANDOM_SEED = 42

perceptron = Perceptron()
perceptron.fit(X_train.values, y_train.values, num_epochs=5000)

sk_perceptron = skPerceptron(random_state=RANDOM_SEED, max_iter=5000)
sk_perceptron.fit(X_train.values, y_train.values, )
```

```python
print('Точность (доля правильных ответов, из 100%) собственного перцептрона: {:.3f} %'.format(
    accuracy_score(y_test.values, perceptron.forward_pass(X_test)) * 100))
print('Точность (доля правильных ответов) перцептрона из sklearn: {:.3f} %'.format(
    accuracy_score(y_test.values, sk_perceptron.predict(X_test)) * 100))
```

### Важно
